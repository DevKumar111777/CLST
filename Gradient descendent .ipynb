```python
def update(y, m, x, b, learning_rate):
    n = len(x)
    y_pred = m * x + b
    error = y - y_pred
    dm = (-2 / n) * np.dot(x, error)
    db = (-2 / n) * np.sum(error)
    m -= learning_rate * dm
    b -= learning_rate * db
    return m, b
```

```python
def gradient_descent(x_rand, y, learning_rate, n_iterations):
    m = 0
    b = 0
    for i in range(1, n_iterations + 1):
        m, b = update(y, m, x_rand, b, learning_rate)
        if i % 100 == 0:
            y_pred = m * x_rand + b
            cost = compute_cost(y, y_pred)
            print(f"Iteration {i}: Cost = {cost:.4f}, Slope = {m:.4f}, Intercept = {b:.4f}")
    return m, b

```

```python
m, b = gradient_descent(x_rand, y, learning_rate, n_iterations)
```
